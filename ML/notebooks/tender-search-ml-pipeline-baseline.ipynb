```python
train_df = pd.read_csv(SPLITS_DIR / 'train_ranker.csv')
val_df = pd.read_csv(SPLITS_DIR / 'val_ranker.csv')
test_df = pd.read_csv(SPLITS_DIR / 'test_ranker.csv')

print('train:', train_df.shape)
print('val:  ', val_df.shape)
print('test: ', test_df.shape)

train_df.head(2)
```

```python
def label_stats(df: pd.DataFrame, name: str) -> None:
    sessions = df['session_id'].nunique()
    avg_candidates = len(df) / sessions
    dist = df['label'].value_counts(normalize=True).sort_index()
    print(f'{name}: sessions={sessions}, avg_candidates={avg_candidates:.2f}')
    print(dist.to_frame('share').T)

label_stats(train_df, 'train')
label_stats(val_df, 'val')
label_stats(test_df, 'test')
```

## Фиче-инженерия

Базовые признаки:
- длина запроса и позиция кандидата;
- совпадение **названия** категории товара с топ-категориями пользователя и организации;
- совпадение **названия** поставщика товара с топ-поставщиками пользователя;
- признаки пользовательской истории (недавние СТЕ, популярные запросы);
- TF-IDF по тексту запроса и атрибутам товара.


```python
def safe_json(value: object) -> object:
    if isinstance(value, (list, dict)):
        return value
    if value is None:
        return []
    text = str(value).strip()
    if not text:
        return []
    try:
        return json.loads(text)
    except json.JSONDecodeError:
        return []

def build_features(df: pd.DataFrame) -> pd.DataFrame:
    x = df.copy()

    # Backward compatibility: older split files may not contain *_name columns.
    if 'item_category_name' not in x.columns:
        x['item_category_name'] = x.get('item_category_id', '').fillna('')
    if 'item_supplier_name' not in x.columns:
        x['item_supplier_name'] = x.get('item_supplier_id', '').fillna('')

    x['item_category_name'] = x['item_category_name'].fillna('')
    x['item_supplier_name'] = x['item_supplier_name'].fillna('')

    x['user_top_categories'] = x['user_top_categories_json'].map(safe_json)
    x['user_recent_ste_ids'] = x['user_recent_ste_ids_json'].map(safe_json)
    x['user_top_suppliers'] = x['user_top_suppliers_json'].map(safe_json)
    x['user_popular_queries'] = x['user_popular_queries_json'].map(safe_json)
    x['org_top_categories'] = x['org_top_categories_json'].map(safe_json)
    x['item_attrs'] = x['item_attributes_json'].map(safe_json)

    x['query_len'] = x['query'].fillna('').str.split().str.len().astype(float)
    x['normalized_query_len'] = x['normalized_query'].fillna('').str.split().str.len().astype(float)
    x['position_num'] = pd.to_numeric(x['position'], errors='coerce').fillna(999).astype(float)

    x['cat_match_user'] = x.apply(
        lambda r: float(r['item_category_name'] in r['user_top_categories']),
        axis=1,
    )
    x['cat_match_org'] = x.apply(
        lambda r: float(r['item_category_name'] in r['org_top_categories']),
        axis=1,
    )
    x['supplier_match_user'] = x.apply(
        lambda r: float(r['item_supplier_name'] in r['user_top_suppliers']),
        axis=1,
    )
    x['is_recent_ste'] = x.apply(
        lambda r: float(r['ste_id'] in r['user_recent_ste_ids']),
        axis=1,
    )
    x['query_in_history'] = x.apply(
        lambda r: float(str(r['query']) in set(r['user_popular_queries'])),
        axis=1,
    )

    x['item_keywords'] = x['item_attrs'].map(
        lambda d: ' '.join(str(v) for v in d.values()) if isinstance(d, dict) else ''
    )
    x['text_input'] = (
        x['query'].fillna('') + ' ' +
        x['normalized_query'].fillna('') + ' ' +
        x['item_category_id'].fillna('') + ' ' +
        x['item_category_name'].fillna('') + ' ' +
        x['item_supplier_id'].fillna('') + ' ' +
        x['item_supplier_name'].fillna('') + ' ' +
        x['item_keywords'].fillna('')
    )
    return x

train_feat = build_features(train_df)
val_feat = build_features(val_df)
test_feat = build_features(test_df)

numeric_cols = [
    'query_len',
    'normalized_query_len',
    'position_num',
    'cat_match_user',
    'cat_match_org',
    'supplier_match_user',
    'is_recent_ste',
    'query_in_history',
]

train_feat[numeric_cols].head(2)
```

```python
vectorizer = TfidfVectorizer(min_df=2, max_features=30000, ngram_range=(1, 2))

X_train_text = vectorizer.fit_transform(train_feat['text_input'])
X_val_text = vectorizer.transform(val_feat['text_input'])
X_test_text = vectorizer.transform(test_feat['text_input'])

X_train_num = csr_matrix(train_feat[numeric_cols].astype(float).values)
X_val_num = csr_matrix(val_feat[numeric_cols].astype(float).values)
X_test_num = csr_matrix(test_feat[numeric_cols].astype(float).values)

X_train = hstack([X_train_text, X_train_num], format='csr')
X_val = hstack([X_val_text, X_val_num], format='csr')
X_test = hstack([X_test_text, X_test_num], format='csr')

y_train = train_feat['label'].astype(float).values
y_val = val_feat['label'].astype(float).values
y_test = test_feat['label'].astype(float).values

model = Ridge(alpha=1.0)
model.fit(X_train, y_train)

val_pred = model.predict(X_val)
test_pred = model.predict(X_test)

val_pred[:5]
```

```python
def dcg_at_k(relevances: np.ndarray, k: int = 10) -> float:
    rel = np.asarray(relevances)[:k]
    if rel.size == 0:
        return 0.0
    denom = np.log2(np.arange(2, rel.size + 2))
    return float(np.sum((2 ** rel - 1) / denom))

def ndcg_at_k(df: pd.DataFrame, pred_col: str, label_col: str = 'label', group_col: str = 'session_id', k: int = 10) -> float:
    scores = []
    for _, grp in df.groupby(group_col):
        rel_true_sorted = grp.sort_values(label_col, ascending=False)[label_col].to_numpy()
        rel_pred_sorted = grp.sort_values(pred_col, ascending=False)[label_col].to_numpy()
        ideal = dcg_at_k(rel_true_sorted, k)
        if ideal == 0:
            continue
        scores.append(dcg_at_k(rel_pred_sorted, k) / ideal)
    return float(np.mean(scores)) if scores else 0.0

val_eval = val_feat[['session_id', 'label']].copy()
val_eval['pred'] = val_pred

test_eval = test_feat[['session_id', 'label']].copy()
test_eval['pred'] = test_pred

val_ndcg10 = ndcg_at_k(val_eval, pred_col='pred', k=10)
test_ndcg10 = ndcg_at_k(test_eval, pred_col='pred', k=10)

print(f'VAL NDCG@10:  {val_ndcg10:.4f}')
print(f'TEST NDCG@10: {test_ndcg10:.4f}')
```

```python
artifacts_dir = (DATA_DIR.parent.parent / 'models' / 'baseline_v1').resolve()
artifacts_dir.mkdir(parents=True, exist_ok=True)

joblib.dump(model, artifacts_dir / 'ridge_ranker.joblib')
joblib.dump(vectorizer, artifacts_dir / 'tfidf_vectorizer.joblib')
(artifacts_dir / 'numeric_features.json').write_text(
    json.dumps(numeric_cols, ensure_ascii=False, indent=2),
    encoding='utf-8',
)

metrics = {
    'val_ndcg10': val_ndcg10,
    'test_ndcg10': test_ndcg10,
    'train_rows': int(train_df.shape[0]),
    'val_rows': int(val_df.shape[0]),
    'test_rows': int(test_df.shape[0]),
}
(artifacts_dir / 'metrics.json').write_text(
    json.dumps(metrics, ensure_ascii=False, indent=2),
    encoding='utf-8',
)

artifacts_dir
```
